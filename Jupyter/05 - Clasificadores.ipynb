```python
dfBikes = pd.read_csv('https:kanaries-app.s3.ap-northeast-1.amazonaws.com/public-datasets/bike_sharing_dc.csv', parse_dates=[])
dfBikes
```

```python
dfBikes.stypes
```

```python
#Etiquetas
estacion = dfBikes.groupby('season').size().index
estacion
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
codificador = LabelEncoder()
dfBikes['season'] = codificador.fit_transform(dfBikes['season'])
dfBikes['holiday'] = codificador.fit_transform(dfBikes['holiday'])
dfBikes['am or pm'] = codificador.fit_transform(dfBikes['am or pm'])
dfBikes
```

```python
train, test = train_test_split(dfBikes, test_size=0.4, stratify=dfBikes['season'], random_state=10)
display(train.info())
```

```python
train.describe()
```

```python
matriz = train.corr()
plt.figure(figsize=(12,8))
sns.heatmap(matriz, annot=True, cbar=False, linewidths=1)
plt.title('Matriz de correlacion')
plt.show()
```

```python
display(train['season'].value_counts())
```

```python
sns.color_palette('hls', 8)
plt.figure(figsize=(8,6))
sns.countplot('season', data=train)
#Extraer el eje del grafico
ax=plt.gca()
#Modificar las dimenciones del grafico
ymax=train['season'].value_counts().max()
ax.set_ylim(0,3000)
#Agregar la etiqueta para la serie de datos x
ax.set_xticklabels(estacion)
#Recorrer x para obtener los valores y modificar el aspecto del grafico
for p in ax.patches():
    ax.text(p.get_x() + p.get_width()/2, p.get_height(), '%d' % int(p.get_height()), fontsize=12, color='black', ha='center', va='botton')
    
```
