```python
from transformers import pipeline
```

```python
clasificador = pipeline('sentiment-analysis')
```

```python
textos = [
    "I loved the product; it exceeded all my expectations.",
    "Customer service was terrible, and shipping took forever.",
    "The quality is decent, but nothing special.",
    "It was exactly what I was looking for.",
    "Everyone should have one."
]
```

```python
# Realizar la predicción
prediccion = clasificador(textos)
```

```python
for textos, resultado in zip(textos, prediccion):
  sentimiento = resultado['label']
  puntuacion = resultado['score']
  print('-'*30)
  print(f'Texto: **{textos}**')
  print(f'Sentimiento: **{sentimiento}**')
  print(f'Confianza (score): **{puntuacion:.4f}**')
```
